## Dataset Exploration of Public Nusaince Analysis 


```python
import cv2
import os
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset_path = "../data/raw"

print("Dataset Path:", dataset_path)
```

```python
videos = []

for root, folders, files in os.walk(dataset_path):
    for file in files:
        if file.endswith((".mp4", ".avi", ".mov", ".mkv")):
            videos.append(os.path.join(root, file))

print("Total Videos:", len(videos))
```

```python
classes = os.listdir(dataset_path)

print("Classes:")
for c in classes:
    print(c)
```

```python
for c in classes:
    class_path = os.path.join(dataset_path, c)

    if os.path.isdir(class_path):
        count = 0

        for file in os.listdir(class_path):
            if file.endswith((".mp4", ".avi", ".mov", ".mkv")):
                count += 1

        print(c, ":", count)
```

```python
video_path = videos[0]

cap = cv2.VideoCapture(video_path)

fps = cap.get(cv2.CAP_PROP_FPS)
frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

duration = frames / fps

print("FPS:", fps)
print("Frames:", frames)
print("Resolution:", width, "x", height)
print("Duration:", duration, "seconds")

cap.release()
```

```python
data = []

for video in videos[:20]:  # initially testing with 20 videos

    cap = cv2.VideoCapture(video)

    fps = cap.get(cv2.CAP_PROP_FPS)
    frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

    if fps > 0:
        duration = frames / fps
    else:
        duration = 0

    data.append([
        video,
        fps,
        frames,
        width,
        height,
        duration
    ])

    cap.release()

df = pd.DataFrame(
    data,
    columns=[
        "Video",
        "FPS",
        "Frames",
        "Width",
        "Height",
        "Duration"
    ]
)

df.head()
```

```python
plt.hist(df["Duration"])

plt.xlabel("Duration (seconds)")
plt.ylabel("Number of Videos")
plt.title("Video Duration Distribution")

plt.show()
```

```python
df.to_csv(
    "../data/processed/video_metadata.csv",
    index=False
)

print("Metadata saved successfully.")
```
