```python
import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO
```

```python
model = YOLO("yolo11n.pt")

print("YOLO model loaded")
```

```python
video_path = "../data/raw/Abuse/sample.mp4"

cap = cv2.VideoCapture(video_path)

print("Video opened:", cap.isOpened())
```

```python
ret, frame = cap.read()

if ret:
    print("Frame successfully read")
```

```python
results = model(frame)

print("Detection completed")
```

```python
detected_frame = results[0].plot()

detected_frame = cv2.cvtColor(
    detected_frame,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(10, 6))
plt.imshow(detected_frame)
plt.title("YOLO Object Detection")
plt.axis("off")
plt.show()
```

```python
for box in results[0].boxes:

    class_id = int(box.cls[0])
    confidence = float(box.conf[0])

    class_name = model.names[class_id]

    print(
        "Object:",
        class_name,
        "| Confidence:",
        round(confidence, 2)
    )
```

```python
cap = cv2.VideoCapture(video_path)

frames = []

for i in range(5):

    ret, frame = cap.read()

    if not ret:
        break

    results = model(frame)

    output = results[0].plot()

    output = cv2.cvtColor(
        output,
        cv2.COLOR_BGR2RGB
    )

    frames.append(output)

cap.release()

print("Processed frames:", len(frames))
```

```python
plt.figure(figsize=(15, 8))

for i, frame in enumerate(frames):

    plt.subplot(2, 3, i + 1)
    plt.imshow(frame)
    plt.title(f"Frame {i + 1}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
confidence_threshold = 0.5

results = model(
    frame,
    conf=confidence_threshold
)

output = results[0].plot()

output = cv2.cvtColor(
    output,
    cv2.COLOR_BGR2RGB
)

plt.imshow(output)
plt.axis("off")
plt.show()
```
