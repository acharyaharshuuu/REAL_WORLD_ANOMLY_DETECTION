## Event Generation & Temporal Smoothing of Public Nusaince Analysis 


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
predictions = [
    ("Assault", 0.42),
    ("Assault", 0.68),
    ("Assault", 0.91),
    ("Assault", 0.94),
    ("Assault", 0.89),
    ("Assault", 0.86),
    ("Normal", 0.40),
    ("Normal", 0.35),
    ("Normal", 0.30)
]

for i, (activity, confidence) in enumerate(predictions):
    print(
        "Clip:", i + 1,
        "| Activity:", activity,
        "| Confidence:", confidence
    )
```

```python
confidence_threshold = 0.70

print("Confidence Threshold:", confidence_threshold)
```

```python
detected = []

for activity, confidence in predictions:

    if confidence >= confidence_threshold:
        detected.append(
            (activity, confidence)
        )

print("Detected Predictions:")

for item in detected:
    print(item)
```

```python
min_consecutive = 3

consecutive_count = 0
event_detected = False

for activity, confidence in predictions:

    if activity == "Assault" and confidence >= confidence_threshold:

        consecutive_count += 1

    else:

        consecutive_count = 0

    if consecutive_count >= min_consecutive:

        event_detected = True

print("Potential Event Detected:", event_detected)
```

```python
event = {
    "activity": "Assault",
    "start_clip": 3,
    "end_clip": 6,
    "peak_confidence": 0.94
}

print(event)
```

```python
clip_duration = 2
```

```python
start_clip = 3
end_clip = 6

start_time = (start_clip - 1) * clip_duration
end_time = end_clip * clip_duration

print("Start Time:", start_time, "seconds")
print("End Time:", end_time, "seconds")
```

```python
def detect_event(predictions, threshold=0.70, min_consecutive=3):

    count = 0
    start = None
    peak_confidence = 0

    for i, (activity, confidence) in enumerate(predictions):

        if activity != "Normal" and confidence >= threshold:

            if count == 0:
                start = i

            count += 1
            peak_confidence = max(
                peak_confidence,
                confidence
            )

        else:

            if count >= min_consecutive:

                return {
                    "start_clip": start + 1,
                    "end_clip": i,
                    "peak_confidence": peak_confidence
                }

            count = 0
            peak_confidence = 0

    return None
```

```python
event = detect_event(
    predictions,
    threshold=0.70,
    min_consecutive=3
)

print(event)
```

```python
confidences = [
    confidence
    for activity, confidence in predictions
]

plt.plot(confidences, marker="o")

plt.axhline(
    confidence_threshold,
    linestyle="--"
)

plt.xlabel("Clip Number")
plt.ylabel("Confidence")
plt.title("Activity Confidence Over Time")

plt.show()
```
