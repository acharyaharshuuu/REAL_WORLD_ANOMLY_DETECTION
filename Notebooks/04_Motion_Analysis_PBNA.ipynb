## Motion Analysis of Public Nusaince Analysis 


```python
import cv2
import matplotlib.pyplot as plt
```

```python
video_path = "../data/raw/Abuse/sample.mp4"

cap = cv2.VideoCapture(video_path)

print("Video opened:", cap.isOpened())
```

```python
ret, frame1 = cap.read()
ret, frame2 = cap.read()

gray1 = cv2.cvtColor(frame1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(frame2, cv2.COLOR_BGR2GRAY)
```

```python
difference = cv2.absdiff(gray1, gray2)

plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
plt.imshow(gray1, cmap="gray")
plt.title("Frame 1")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(gray2, cmap="gray")
plt.title("Frame 2")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(difference, cmap="gray")
plt.title("Frame Difference")
plt.axis("off")

plt.show()
```

```python
_, threshold = cv2.threshold(
    difference,
    30,
    255,
    cv2.THRESH_BINARY
)

plt.imshow(threshold, cmap="gray")
plt.title("Motion after Thresholding")
plt.axis("off")
plt.show()
```

```python
cap = cv2.VideoCapture(video_path)

background = cv2.createBackgroundSubtractorMOG2()

ret, frame = cap.read()

mask = background.apply(frame)

plt.imshow(mask, cmap="gray")
plt.title("Background Subtraction")
plt.axis("off")
plt.show()
```

```python
for i in range(20):

    ret, frame = cap.read()

    if not ret:
        break

    mask = background.apply(frame)

plt.imshow(mask, cmap="gray")
plt.title("Detected Motion")
plt.axis("off")
plt.show()

cap.release()
```

```python
cap = cv2.VideoCapture(video_path)

ret, first_frame = cap.read()
ret, second_frame = cap.read()

old_gray = cv2.cvtColor(first_frame, cv2.COLOR_BGR2GRAY)
new_gray = cv2.cvtColor(second_frame, cv2.COLOR_BGR2GRAY)

cap.release()
```

```python
flow = cv2.calcOpticalFlowFarneback(
    old_gray,
    new_gray,
    None,
    0.5,
    3,
    15,
    3,
    5,
    1.2,
    0
)

magnitude, angle = cv2.cartToPolar(
    flow[..., 0],
    flow[..., 1]
)

plt.imshow(magnitude, cmap="gray")
plt.title("Optical Flow Magnitude")
plt.colorbar()
plt.show()
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(difference, cmap="gray")
plt.title("Frame Difference")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(threshold, cmap="gray")
plt.title("Threshold")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(magnitude, cmap="gray")
plt.title("Optical Flow")
plt.axis("off")

plt.tight_layout()
plt.show()
```
