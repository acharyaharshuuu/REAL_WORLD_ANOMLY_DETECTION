## Frame Sampling & Clipping of Public Nusaince Analysis 


```python
import cv2
import matplotlib.pyplot as plt
```

```python
video_path = "../data/raw/Abuse/sample.mp4"

cap = cv2.VideoCapture(video_path)

print("Video opened:", cap.isOpened())
```

```python
fps = cap.get(cv2.CAP_PROP_FPS)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

print("FPS:", fps)
print("Total Frames:", total_frames)
```

```python
sample_rate = 10

sampled_frames = []

frame_number = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    if frame_number % sample_rate == 0:

        frame = cv2.resize(frame, (640, 360))
        frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

        sampled_frames.append(frame)

    frame_number += 1

cap.release()

print("Sampled Frames:", len(sampled_frames))
```

```python
plt.figure(figsize=(15, 6))

for i in range(min(6, len(sampled_frames))):

    plt.subplot(2, 3, i + 1)

    plt.imshow(sampled_frames[i])
    plt.title(f"Sample {i + 1}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
clip_size = 16

clip = sampled_frames[:clip_size]

print("Frames in clip:", len(clip))
```

```python
plt.figure(figsize=(16, 8))

for i, frame in enumerate(clip):

    plt.subplot(4, 4, i + 1)

    plt.imshow(frame)
    plt.title(f"Frame {i + 1}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
def sample_frames(video_path, sample_rate=10):

    cap = cv2.VideoCapture(video_path)

    frames = []
    frame_number = 0

    while True:

        ret, frame = cap.read()

        if not ret:
            break

        if frame_number % sample_rate == 0:

            frame = cv2.resize(frame, (640, 360))
            frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

            frames.append(frame)

        frame_number += 1

    cap.release()

    return frames
```

```python
frames = sample_frames(
    video_path,
    sample_rate=10
)

print("Total sampled frames:", len(frames))
```

```python
clip_size = 16

clips = []

for i in range(0, len(frames), clip_size):

    clip = frames[i:i + clip_size]

    if len(clip) == clip_size:
        clips.append(clip)

print("Total Clips:", len(clips))
```
