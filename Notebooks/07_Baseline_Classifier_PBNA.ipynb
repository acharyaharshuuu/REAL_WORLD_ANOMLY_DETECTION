## Baseline Classifier of Public Nusaince Analysis 


```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
X = np.array([
    [0.12, 0.45, 0.32, 0.67],
    [0.21, 0.34, 0.56, 0.71],
    [0.78, 0.23, 0.11, 0.42],
    [0.81, 0.31, 0.15, 0.38],
    [0.14, 0.51, 0.36, 0.69],
    [0.76, 0.25, 0.18, 0.41]
])

y = np.array([
    "Abuse",
    "Abuse",
    "Assault",
    "Assault",
    "Abuse",
    "Assault"
])

print("Feature shape:", X.shape)
print("Labels:", y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
classifier = LogisticRegression()

classifier.fit(
    X_train,
    y_train
)

print("Model trained")
```

```python
predictions = classifier.predict(X_test)

print("Predictions:")
print(predictions)
```

```python
accuracy = accuracy_score(
    y_test,
    predictions
)

print("Accuracy:", accuracy)
```

```python
print(
    classification_report(
        y_test,
        predictions,
        zero_division=0
    )
)
```
